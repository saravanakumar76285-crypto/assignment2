1. Print numbers from 1 to 10 using a for loop

2. Print numbers from 10 to 1 using a while loop. 

```python
i=10
while i>0:
    print(i)
    i -= 1
```

3. Write a program to print the multiplication table of a given number. 

```python
num = int(input("Enter a number: "))
for i in range(1,11):
    print(i,"x",num,"=",i*num)
```

4.Write a program to find the sum of first N natural numbers using a loop. 

```python
num=int(input("Enter a N number: "))
total=0
for i in range(1,num+1):
    total=total+i
print("The sum of first",num,"numbers is:",total)
```

8. Write a function is_even(n) that returns True if the number is even, otherwise False. 

```python
n=int(input("enter a number:"))
def even(n):
  if n%2==0:
    return True
  else:
    return False
num=even(n)
print(num)
```

10. Write a function max_of_two(a, b) that returns the larger of two numbers 

```python
a=int(input("enter a:"))
b=int(input("enter b:"))
def max_of_two(a,b):
  if a>b:
    return a
  else:
    return b
max=max_of_two(a,b)
print("the largest is:",max)
```
